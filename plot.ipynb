```python
import matplotlib.pyplot as plt
import pandas as pd

# load data
df = pd.read_csv("istherecorrelation.csv", sep=";", decimal=",")

# figure resolution
plt.rcParams["figure.dpi"] = 300

fig, ax1 = plt.subplots(figsize=(9, 6))

# left y axis: university students
color_wo = "blue"
ax1.set_xlabel("Year", fontsize=11)
ax1.set_ylabel("WO Students (x1000)", color=color_wo, fontsize=11)
(line1,) = ax1.plot(
    df["Year"],
    df["WO [x1000]"],
    color=color_wo,
    marker="o",
    linewidth=2,
    label="WO Students",
)
ax1.tick_params(axis="y", labelcolor=color_wo)
ax1.set_ylim(bottom=0)  # start left y-axis at 0
ax1.grid(True, linestyle="--", alpha=0.4)

# right y-axis: beer consumption
ax2 = ax1.twinx()
color_beer = "red"
ax2.set_ylabel(
    "Beer Consumption (x1000 hectoliters)", color=color_beer, fontsize=11
)
(line2,) = ax2.plot(
    df["Year"],
    df["NL Beer consumption [x1000 hectoliter]"],
    color=color_beer,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Beer Consumption",
)
ax2.tick_params(axis="y", labelcolor=color_beer)
ax2.set_ylim(bottom=0)  # start right y-axis at 0

# single combined legend
lines = [line1, line2]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc="lower right")

# ensure year ticks remain whole numbers
ax1.set_xticks(df["Year"])

plt.title(
    "WO Student Enrollment vs. Dutch Beer Consumption (200-2018)", pad=12
)
fig.tight_layout()

plt.savefig("graph.png", bbox_inches="tight")
plt.show()
```
